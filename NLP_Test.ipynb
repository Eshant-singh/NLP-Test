# NLP Introduction and Text Preprocessing

## Question 1

**What is the primary goal of Natural Language Processing (NLP)?**

The primary goal of Natural Language Processing is to enable computers to understand, process, and work with human language. NLP helps machines analyze text and speech so that useful information can be extracted and language-based tasks can be performed.

## Question 2

**What does "tokenization" refer to in text processing?**

Tokenization is the process of breaking text into smaller units called tokens. These tokens can be words, sentences, or other meaningful parts of text. For example, a sentence can be split into individual words.

## Question 3

**What is the difference between lemmatization and stemming?**

Stemming removes prefixes or suffixes from words to obtain a shorter root form, which may not always be a valid word. Lemmatization uses linguistic information to convert a word into its meaningful base or dictionary form. Therefore, lemmatization is generally more accurate but can take more processing time.

## Question 4

**What is the role of regular expressions (regex) in text processing?**

Regular expressions are patterns used to search, match, and manipulate text. In NLP, regex can be used to remove unwanted characters, extract emails or numbers, find specific patterns, and clean text before further processing.

## Question 5

**What is Word2Vec and how does it represent words in a vector space?**

Word2Vec is a word embedding technique that learns numerical representations of words from their surrounding context. Each word is represented as a vector in a continuous vector space. Words used in similar contexts tend to have similar vector representations.

## Question 6

**How does frequency distribution help in text analysis?**

Frequency distribution counts how often words or other tokens occur in a text. It helps identify common and important terms, understand the vocabulary of a document, and find patterns in the data.

## Question 7

**Why is text normalization important in NLP?**

Text normalization converts different forms of text into a more consistent format. It can include converting text to lowercase, removing punctuation, handling extra spaces, and applying other standardization steps. This makes later NLP processing more consistent.

## Question 8

**What is the difference between sentence tokenization and word tokenization?**

Sentence tokenization divides a document into individual sentences, while word tokenization divides a sentence or document into individual words or tokens. Sentence tokenization works at the sentence level, whereas word tokenization works at the word level.

## Question 9

**What are co-occurrence vectors in NLP?**

Co-occurrence vectors represent words using information about the words that appear near them. A vector can contain counts showing how frequently a target word occurs with other words in a context. These vectors can be used to study relationships between words.

## Question 10

**What is the significance of lemmatization in improving NLP tasks?**

Lemmatization reduces different forms of a word to a common meaningful base form. This can reduce unnecessary variation in text and help NLP systems treat related word forms as the same concept, improving tasks such as search, classification, and text analysis.

## Question 11

**What is the primary use of word embeddings in NLP?**

The primary use of word embeddings is to represent words as numerical vectors that capture useful relationships and similarities between words. These representations can then be given as input to machine learning and deep learning models.

## Question 12

**What is an annotator in NLP?**

An annotator is a person or system that adds labels or linguistic information to text. For example, an annotator may identify parts of speech, named entities, sentiment, or other properties in a dataset.

## Question 13

**What are the key steps in text processing before applying machine learning models?**

Common steps include collecting the text, cleaning and normalizing it, tokenizing the text, removing unwanted words or characters, applying stemming or lemmatization when required, and converting the processed text into numerical features such as vectors or embeddings.

## Question 14

NLP initially relied heavily on manually written rules and linguistic approaches. Later, statistical methods and machine learning became common. With the growth of deep learning, neural networks and word embeddings became widely used, followed by transformer-based models that have greatly improved many language tasks.

## Question 15

**Why is sentence processing important in NLP?**

Sentence processing helps NLP systems understand the boundaries and structure of sentences. It is an important preprocessing step because many later tasks, such as parsing, sentiment analysis, and information extraction, depend on correctly identifying individual sentences.

## Question 16

**How do word embeddings improve the understanding of language semantics in NLP?**

Word embeddings represent words as vectors learned from their contexts. Words with similar meanings or usage patterns often have similar vector representations. This allows machine learning models to capture semantic relationships more effectively than simple word counts.

## Question 17

## Question 18

## Question 21

## Question 3

**How can you remove stopwords from a sentence?**

```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download("punkt")
nltk.download("stopwords")

text = "This is a simple example of natural language processing."

words = word_tokenize(text)
stop_words = set(stopwords.words("english"))

filtered_words = [
    word for word in words
    if word.lower() not in stop_words
]

print(filtered_words)
```

## Question 4

**How can you perform stemming on a word?**

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

word = "running"

stemmed_word = stemmer.stem(word)

print("Original:", word)
print("Stemmed:", stemmed_word)
```

## Question 5

**How can you perform lemmatization on a word?**

```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet")

lemmatizer = WordNetLemmatizer()

word = "running"

lemma = lemmatizer.lemmatize(word, pos="v")

print("Original:", word)
print("Lemma:", lemma)
```

## Question 7

**How can you create a co-occurrence matrix for words in a corpus?**

```python
import numpy as np
from collections import Counter

corpus = [
    "cat likes milk",
    "cat likes fish",
    "dog likes fish"
]

words = sorted(set(" ".join(corpus).split()))
word_index = {word: i for i, word in enumerate(words)}

matrix = np.zeros((len(words), len(words)), dtype=int)

for sentence in corpus:
    tokens = sentence.split()

    for i, word in enumerate(tokens):
        for j, context_word in enumerate(tokens):
            if i != j:
                matrix[word_index[word], word_index[context_word]] += 1

print("Words:", words)
print(matrix)
```

## Question 8

**How can you apply a regular expression to extract all email addresses from a text?**

```python
import re

text = "Contact us at abc@gmail.com or support@example.com."

pattern = r"[\w.-]+@[\w.-]+\.\w+"

emails = re.findall(pattern, text)

print(emails)
```

## Question 9

**How can you perform word embedding using Word2Vec?**

```python
from gensim.models import Word2Vec

sentences = [
    ["natural", "language", "processing"],
    ["machine", "learning", "is", "useful"],
    ["natural", "language", "is", "useful"]
]

model = Word2Vec(
    sentences,
    vector_size=50,
    window=3,
    min_count=1,
    workers=1
)

print(model.wv["language"])
```

## Question 10

**How can you use Doc2Vec to embed documents?**

```python
from gensim.models.doc2vec import Doc2Vec, TaggedDocument

documents = [
    TaggedDocument(
        words=["natural", "language", "processing"],
        tags=["doc1"]
    ),
    TaggedDocument(
        words=["machine", "learning", "is", "useful"],
        tags=["doc2"]
    )
]

model = Doc2Vec(
    documents,
    vector_size=50,
    min_count=1,
    epochs=20
)

vector = model.dv["doc1"]

print(vector)
```

## Question 12

**How can you find the similarity between two sentences using cosine similarity?**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

sentences = [
    "I like machine learning",
    "I enjoy machine learning"
]

vectorizer = TfidfVectorizer()
vectors = vectorizer.fit_transform(sentences)

similarity = cosine_similarity(vectors[0], vectors[1])

print("Cosine Similarity:", similarity[0][0])
```

## Question 14

**How can you split a large document into smaller chunks of text?**

```python
text = """Natural Language Processing is a field of artificial intelligence.
It helps computers work with human language.
Text preprocessing is an important part of NLP."""

chunk_size = 80

chunks = [
    text[i:i + chunk_size]
    for i in range(0, len(text), chunk_size)
]

for i, chunk in enumerate(chunks, 1):
    print("Chunk", i)
    print(chunk)
    print()
```

## Question 15

**How can you calculate the TF-IDF (Term Frequency - Inverse Document Frequency) for a set of documents?**

```python
from sklearn.feature_extraction.text import TfidfVectorizer

documents = [
    "machine learning is useful",
    "natural language processing is useful",
    "machine learning and NLP are interesting"
]

vectorizer = TfidfVectorizer()

tfidf_matrix = vectorizer.fit_transform(documents)

print("Features:")
print(vectorizer.get_feature_names_out())

print("\nTF-IDF Matrix:")
print(tfidf_matrix.toarray())
```

## Question 16

**How can you apply tokenization, stopword removal, and stemming in one go?**

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer

nltk.download("punkt")
nltk.download("stopwords")

text = "Natural language processing is processing language."

words = word_tokenize(text)
stop_words = set(stopwords.words("english"))
stemmer = PorterStemmer()

processed_words = []

for word in words:
    if word.isalpha() and word.lower() not in stop_words:
        processed_words.append(stemmer.stem(word.lower()))

print(processed_words)
```

## Question 17

**How can you visualize the frequency distribution of words in a sentence?**

```python
import nltk
import matplotlib.pyplot as plt
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist

nltk.download("punkt")

text = "NLP is useful and NLP is interesting."

words = word_tokenize(text.lower())
frequency = FreqDist(words)

frequency.plot(10, title="Word Frequency Distribution")
plt.show()
```
